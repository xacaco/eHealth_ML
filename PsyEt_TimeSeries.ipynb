## Seasonal-Trend Decomposition

### Imports

```python
from statsmodels.tsa.seasonal import STL
from statsmodels.tsa.seasonal import seasonal_decompose
import itertools
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import STL
import matplotlib.dates as mdates
```

### Session counts

#### Using STL

```python
# Apply STL decomposition
stl = STL(sessions_per_week['session_count'], period=52)  # ~52 weeks in a year
result = stl.fit()

# Plot the STL components
plt.figure(figsize=(14, 8))
result.plot()
plt.suptitle('STL Decomposition of Weekly Session Counts', fontsize=16)
plt.tight_layout()
plt.show()
```

#### Using Multiplicative STL

```python
# Seasonal trend decomposition
decomposition_sessions = seasonal_decompose(sessions_per_week['session_count'], model='multiplicative', period=52)

# Plot the decomposition components
fig = decomposition_sessions.plot()
fig.suptitle("Time Series Decomposition for Weekly Session Counts", fontsize=16)
plt.tight_layout()
plt.subplots_adjust(top=0.88)  # Adjust to make room for the title
plt.show()
```

```python
# Log-transform for multiplicative STL
log_sessions = np.log(sessions_per_week['session_count'])

# Apply robust STL decomposition
stl = STL(log_sessions, seasonal=53, robust=True)
result = stl.fit()

# Convert components back to multiplicative scale
trend = np.exp(result.trend)
seasonal = np.exp(result.seasonal)
resid = np.exp(result.resid)
observed = sessions_per_week['session_count']

# Plot the decomposition
fig, axs = plt.subplots(4, 1, figsize=(12, 8), sharex=True)
axs[0].plot(observed, label='Observed')
axs[0].legend()
axs[1].plot(trend, label='Trend', color='orange')
axs[1].legend()
axs[2].plot(seasonal, label='Seasonal', color='green')
axs[2].legend()
axs[3].plot(resid, label='Residual', color='red')
axs[3].legend()
plt.tight_layout()
plt.show()
```

```python
# STL decomposition
decomposition_sessions = seasonal_decompose(
    sessions_per_week['session_count'], model='multiplicative', period=52,
)

# Plot decomposition
fig = decomposition_sessions.plot()

# Set figure and axes background color
fig.set_facecolor('#D1F4CA')
for ax in fig.axes:
    ax.set_facecolor('#D1F4CA')
    
    # Set all lines and markers to the new color
    for line in ax.get_lines():
        line.set_color('#000091')  # Deep blue
        line.set_markerfacecolor('#000091')
        line.set_markeredgecolor('#000091')

# Set main title
fig.suptitle("STL on Weekly Session Counts", fontsize=16, fontweight='bold')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
# Extract the seasonal component
seasonal_component = decomposition_sessions.seasonal

# Extract the first 52 weeks (1 full cycle)
seasonal_one_cycle = seasonal_component.iloc[:52]

# Extract the corresponding dates
dates_one_cycle = seasonal_one_cycle.index

# Create month labels from dates
month_labels = pd.to_datetime(dates_one_cycle).to_series().dt.strftime('%b').unique()

# Find first date of each month in the cycle for correct positioning
month_positions = []
seen_months = set()
for i, date in enumerate(dates_one_cycle):
    month = date.strftime('%b')
    if month not in seen_months:
        seen_months.add(month)
        month_positions.append(i)

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(seasonal_one_cycle.values, marker='o')
plt.title('Seasonal Component (One Year Cycle)')
plt.xlabel('Month')
plt.ylabel('Seasonality')
plt.xticks(month_positions, month_labels)
plt.grid(True)
plt.show()
```

## Prophet Model

### Imports

```python
import numpy as np
import pandas as pd
import logging
from prophet import Prophet
from prophet.plot import add_changepoints_to_plot
from scipy.stats import norm
from prophet.diagnostics import cross_validation, performance_metrics
```

### Data preparation

```python
# Ensure the session date is in datetime format
sess['Date de la séance'] = pd.to_datetime(sess['Date de la séance'])

# Function to count distinct psychologists in the 30-day window (including current date)
def count_unique_psychologists(date):
    window_start = date - pd.Timedelta(days=60)
    mask = (sess['Date de la séance'] >= window_start) & (sess['Date de la séance'] <= date)
    return sess.loc[mask, 'psychologistId ID'].nunique()

# Add new column to existing sessions_per_day DataFrame
sessions_per_day['unique_psychologists_active'] = sessions_per_day.index.to_series().apply(count_unique_psychologists)
```

```python
# Step 1: Count sessions per psychologist per day
session_counts = sess.groupby(["psychologistId ID", "Date de la séance"]).size().reset_index(name="daily_sessions")

# Step 2: Aggregate statistics per psychologist
psychologist_stats = session_counts.groupby("psychologistId ID")["daily_sessions"].agg(
    max_sessions_per_day="max",
    min_sessions_per_day="min",
    mean_sessions_per_day="mean"
).reset_index()

psy_df = psy.merge(psychologist_stats, left_on="Dossier Number", right_on="psychologistId ID", how="left")

min_of_mins = psy_df.loc[psy_df["min_sessions_per_day"] > 0, "min_sessions_per_day"].min()
max_of_maxs = psy_df.loc[psy_df["max_sessions_per_day"] > 0, "max_sessions_per_day"].max()
mean_of_means = psy_df.loc[psy_df["mean_sessions_per_day"] > 0, "mean_sessions_per_day"].mean()
```

### Baseline model

```python
# We create 'ds' and 'y' variables
df_prophet = sessions_per_day.reset_index().rename(columns={'Date de la séance':'ds', 'session_count':'y','cap':'cap'})
#df_prophet['cap']=np.log1p(df_prophet['cap'])
#df_prophet['y'] = np.log1p(df_prophet['y'])
df_prophet['floor']=0
#df_prophet['post_2024'] = (df_prophet['ds'] >= '2024-07-01').astype(int)
print(df_prophet.tail())

# Instantiate model
model = Prophet(weekly_seasonality=True,seasonality_mode='multiplicative', growth="linear")
model.add_country_holidays(country_name='FR')
#model.add_regressor('post_2024')
model.fit(df_prophet)
#print(model.train_holiday_names)

# Forecast next 50 weeks
future = model.make_future_dataframe(periods=182, freq='D')
# Merge 'cap' values into 'future'
cap_value = df_prophet['cap'].max()
#future['cap'] = future['cap'].fillna(cap_value)
#future['cap']=cap_value
#future['cap'] = df_prophet['cap'].fillna(cap_value)
future['cap']=cap_value
#future['post_2024'] = (future['ds'] >= '2024-07-01').astype(int)

forecast = model.predict(future)
#df_prophet['y'] = np.expm1(df_prophet['y'])
forecast['yhat_upper'] = np.clip(forecast['yhat_upper'], 0, None)
forecast['yhat_lower'] = np.clip(forecast['yhat_lower'], 0, None)
forecast['yhat'] = np.clip(forecast['yhat'], 0, None)

# Plot prediction
fig = model.plot(forecast)
a = add_changepoints_to_plot(fig.gca(), model, forecast)
fig = model.plot_components(forecast)
plt.show()
```

### Hyperparameter tuning

```python
logging.getLogger("cmdstanpy").disabled = True

# Cross-validation
from prophet.diagnostics import cross_validation

df_cv = cross_validation(model, initial=' 730 days',
    period='15 days',
    horizon='90 days')

# Clip predictions so they don’t go below 0
#df_cv['yhat'] = np.clip(df_cv['yhat'], 0, None)
#df_cv['yhat_lower'] = np.clip(df_cv['yhat_lower'], 0, None)
#df_cv['yhat_upper'] = np.clip(df_cv['yhat_upper'], 0, None)


from prophet.diagnostics import performance_metrics
df_p = performance_metrics(df_cv)
print(df_p)
```

```python
df_cv['y'] = np.expm1(df_cv['y'])
# Step 4. Bias-corrected mean and intervals on original scale
df_cv['yhat'] = np.expm1(df_cv['yhat'])             # corrected mean
df_cv['yhat_lower'] = np.expm1(df_cv['yhat_lower'])            # lower bound
df_cv['yhat_upper'] = np.expm1(df_cv['yhat_upper'])           # upper bound
print(df_cv)

df_p=performance_metrics(df_cv)
print(df_p)
```

```python
logging.getLogger("cmdstanpy").disabled = True

param_grid = {  
    'seasonality_prior_scale': [0.01, 0.1, 1.0, 10.0],
    'changepoint_prior_scale': [0.001, 0.01, 0.1, 0.5],
    'holidays_prior_scale': [0.01, 0.1, 1.0, 10.0]
}

# Generate all combinations of parameters
all_params = [dict(zip(param_grid.keys(), v)) for v in itertools.product(*param_grid.values())]
mapes = []  # Store the MAPEs for each params here

# Use cross validation to evaluate all parameters
for params in all_params:
    m = Prophet(seasonality_mode='additive', growth="linear",**params)
    m.add_country_holidays(country_name='FR')
    m.fit(df_prophet)
    df_cv = cross_validation(m, horizon = '90 days', initial='730 days', period='15 days', parallel="processes")
    df_p = performance_metrics(df_cv, rolling_window=1)
    mapes.append(df_p['mape'].values[0])

# Find the best parameters
tuning_results = pd.DataFrame(all_params)
tuning_results['mape'] = mapes
print(tuning_results)
```

```python
logging.getLogger("cmdstanpy").disabled = True

# Cross-validation
df_cv = cross_validation(model, horizon='30 days')

# --- Bias correction for log-transformed model ---
# Step 1. Check whether df_cv['y'] is still log1p(y) or already original
# If you fitted the model on log1p(y), Prophet thinks that's the "real" y.
# So we need to invert it for evaluation.
df_cv['y'] = np.expm1(df_cv['y'])

# Step 2. Extract mu and upper from log-scale forecasts
mu = df_cv['yhat'].values
upper = df_cv['yhat_upper'].values

# Interval width used in the model (default 0.8 unless changed)
interval_width = model.interval_width
z = norm.ppf((1 + interval_width) / 2.0)

# Step 3. Estimate sigma on the log scale
sigma = (upper - mu) / z
sigma = np.maximum(sigma, 1e-8)  # avoid negatives or zeros

# Step 4. Bias-corrected mean and intervals on original scale
df_cv['yhat'] = np.expm1(mu + 0.5 * sigma**2)             # corrected mean
df_cv['yhat_lower'] = np.expm1(mu - z * sigma)            # lower bound
df_cv['yhat_upper'] = np.expm1(mu + z * sigma)            # upper bound

print(df_cv)

# Performance metrics now use corrected forecasts
df_p = performance_metrics(df_cv)
print(df_p)
```
